```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A_expected_r = 0.05
A_volatility = 0.1

B_expected_r = 0.1
B_volatility = 0.18

corr=0.2

sigma = np.array([[A_volatility**2, corr * A_volatility * B_volatility],
                   [corr * A_volatility * B_volatility, B_volatility**2]])

returns=[]
volatilities=[]

weights = np.linspace(-1, 2, 500)

for w1 in weights:
    w2 = 1 - w1 
    w = np.array([w1, w2])


    P_expected_r = w1 * A_expected_r + w2 * B_expected_r

    
    P_volatility = np.sqrt( np.dot( w.T, np.dot( sigma, w ) ) )
    
    returns += [P_expected_r]
    volatilities += [P_volatility]

    

plt.scatter(volatilities, returns, c='blue')
plt.xlabel('Volatility')
plt.ylabel('Expected return')
plt.grid()
plt.legend()
plt.show()
```

```python

min_volatility = min(volatilities)
index_min = np.argmin(volatilities)
min_w1 = weights[index_min]
min_w2 = 1 - min_w1

print(f"Minimum volatility: {min_volatility:.4f} \nWeights: A:{min_w1:.2f}, B:{min_w2:.2f}")
```

```python
A_expected_r = 0.05
A_volatility = 0.1

B_expected_r = 0.1
B_volatility = 0.18

C_expected_r = 0.06
C_volatility = 0.13

corr=0.2

sigma = np.array([[A_volatility**2, corr * A_volatility * B_volatility, corr * A_volatility * C_volatility],
                   [corr * A_volatility * B_volatility, B_volatility**2, corr * B_volatility * C_volatility], 
                   [corr * A_volatility * C_volatility, corr * B_volatility * C_volatility, C_volatility**2]])  
                



returns=[]
volatilities=[]

weights_range = np.linspace(-1, 2, 100)
weights = []

for w1 in weights_range:
    for w2 in weights_range:

        w3 = 1 - w1 - w2

        if not -1 <= w3 <= 2:
            continue

        w = np.array([w1, w2, w3])


        P_expected_r = w1 * A_expected_r + w2 * B_expected_r + w3 * C_expected_r


        
        P_volatility = np.sqrt( np.dot( w.T, np.dot( sigma, w ) ) )
        
        returns += [P_expected_r]
        volatilities += [P_volatility]
        weights += [w]

plt.scatter(volatilities, returns, c='blue', s=1)
plt.xlabel('Volatility')
plt.ylabel('Expected Return')
plt.grid()
plt.legend()
plt.show()
```
